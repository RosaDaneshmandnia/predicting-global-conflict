```python
import pandas as pd

df = pd.read_csv("../../data/processed/model_data.csv")
print([c for c in df.columns if "holiday" in c.lower()])
```

```python
df["holiday_count_month"].value_counts().sort_index()
```

```python
df["is_holiday_month"].value_counts()
```

```python
df.groupby("is_holiday_month")[[
    "Battles",
    "Explosions/Remote violence",
    "Violence against civilians"
]].mean()
```

```python
df.groupby("holiday_count_month")[[
    "Battles",
    "Explosions/Remote violence",
    "Violence against civilians"
]].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1,3, figsize=(15,4))

sns.boxplot(data=df, x="is_holiday_month", y="Battles", ax=axes[0])
axes[0].set_title("Battles")

sns.boxplot(data=df, x="is_holiday_month", y="Explosions/Remote violence", ax=axes[1])
axes[1].set_title("Explosions/Remote violence")

sns.boxplot(data=df, x="is_holiday_month", y="Violence against civilians", ax=axes[2])
axes[2].set_title("Violence against civilians")

plt.tight_layout()
plt.show()
```
